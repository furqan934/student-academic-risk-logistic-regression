# Student Academic Risk Prediction Using Logistic Regression

## Project Goal

The goal of this project is to develop an interpretable machine learning
system that predicts whether a student is academically at risk using
student demographic, academic, behavioral, and socioeconomic information.

The project will focus on Logistic Regression and will study:

- The mathematical foundation of Logistic Regression
- Implementation from scratch using NumPy
- Data preprocessing
- Gradient descent
- Learning rate and convergence
- Regularization
- Hyperparameter tuning
- Classification thresholds
- Model evaluation
- Coefficient interpretation
- Error analysis
- End-to-end deployment

## Research Perspective

The project will not only focus on achieving high predictive performance.
It will investigate how different modeling decisions affect the behavior,
performance, and interpretability of Logistic Regression.
